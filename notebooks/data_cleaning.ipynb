```python
df = df.drop_duplicates()
```

```python
df.columns = (
    df.columns
      .str.strip()
      .str.lower()
      .str.replace(" ", "_")
)

df.columns
```

```python
df.rename(columns={"sub-category":"sub_category"}, inplace=True)
```

```python
df["order_date"] = pd.to_datetime(
    df["order_date"],
    dayfirst=True
)

df["ship_date"] = pd.to_datetime(
    df["ship_date"],
    dayfirst=True
)
```

```python
missing_percent = (df.isnull().sum()/len(df)*100).round(2)

missing_percent.sort_values(ascending=False)
```

```python
categorical_columns = [
    "region",
    "state",
    "city",
    "category",
    "sub_category",
    "segment",
    "ship_mode"
]

for col in categorical_columns:
    print(f"\n{col}")
    print("-"*50)
    print(df[col].nunique())
    print(df[col].unique())
```

```python
df["year"] = df["order_date"].dt.year
```

```python
df["month"] = df["order_date"].dt.month
```

```python
df["month_name"] = df["order_date"].dt.month_name()
```

```python
df["quarter"] = df["order_date"].dt.quarter
```

```python
df["weekday"] = df["order_date"].dt.day_name()
```

```python
df["shipping_days"] = (
    df["ship_date"] - df["order_date"]
).dt.days
```

```python
order_sales = (
    df.groupby("order_id")["sales"]
    .sum()
)

average_order_value = order_sales.mean()

print(f"Average Order Value : ${average_order_value:.2f}")
```

```python
customer_summary = (
    df.groupby("customer_name")
    .agg(
        Total_Sales=("sales","sum"),
        Orders=("order_id","nunique")
    )
    .reset_index()
)

customer_summary.head()
```

```python
city_summary = (
    df.groupby(["city","state"])
    .agg(
        Total_Sales=("sales","sum"),
        Orders=("order_id","nunique"),
        Customers=("customer_id","nunique"),
        Average_Sale=("sales","mean")
    )
    .reset_index()
)

city_summary.head()
```

```python
city_summary["Demand_Score"] = (
    city_summary["Total_Sales"]
    *
    city_summary["Customers"]
) / city_summary["Orders"]
```

```python
city_summary["Sales_Rank"] = city_summary["Total_Sales"].rank(pct=True)

city_summary["Customer_Rank"] = city_summary["Customers"].rank(pct=True)

city_summary["AOV_Rank"] = city_summary["Average_Sale"].rank(pct=True)
```

```python
city_summary["Opportunity_Score"] = (
    0.5*city_summary["Sales_Rank"] +
    0.3*city_summary["Customer_Rank"] +
    0.2*city_summary["AOV_Rank"]
)
```

```python
top_cities = city_summary.sort_values(
    "Opportunity_Score",
    ascending=False
)

top_cities.head(15)
```

```python
import os

os.makedirs("../data/processed", exist_ok=True)

df.to_csv(
    "../data/processed/cleaned_sales.csv",
    index=False
)

city_summary.to_csv(
    "../data/processed/city_summary.csv",
    index=False
)

customer_summary.to_csv(
    "../data/processed/customer_summary.csv",
    index=False
)

print("Files saved successfully!")
```
